# Benchmark: 完全平均资源分配

本 notebook 计算固定平均资源分配下的独立模块基准。当前 active modules 为 planting 和 mining；加入 fishing 后只修改 `ACTIVE_MODULES`。

## 已确认口径

每个有效模块独立获得相同份额的初始现金、每日时间和每日体力。模块之间不转移现金、矿物或金属锭；每个模块内部约束仍完整保留。详细定义见 `README.md`。

```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent
if not (PROJECT_ROOT / 'planting').exists():
    PROJECT_ROOT = next(p for p in Path.cwd().resolve().parents if (p / 'planting').exists())

ACTIVE_MODULES = ['planting', 'mining']
HORIZON_DAYS = 112
INITIAL_CASH = 500.0
DAILY_TIME = 900.0
DAILY_ENERGY = 270.0

module_count = len(ACTIVE_MODULES)
allocation = pd.DataFrame([{
    'module': module,
    'initial_cash_g': INITIAL_CASH / module_count,
    'daily_time_minutes': DAILY_TIME / module_count,
    'daily_energy': DAILY_ENERGY / module_count,
} for module in ACTIVE_MODULES])
display(allocation)
```
