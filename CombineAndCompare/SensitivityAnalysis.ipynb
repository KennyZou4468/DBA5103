# Integrated Model Sensitivity Analysis

This notebook performs one-at-a-time sensitivity analysis for the integrated planting--mining--fishing MILP. It varies only assumptions that are decision settings rather than fixed game coefficients: initial cash, daily time, daily energy, and Blacksmith service capacity.

Each scenario is solved independently. Because this is a MIP, the main evidence is the change in feasible terminal cash and the reported MIP gap; LP shadow prices are not generally meaningful for the integer model.

```python
from pathlib import Path
import sys

CURRENT_DIR = Path.cwd().resolve()
PROJECT_ROOT = next(p for p in (CURRENT_DIR, *CURRENT_DIR.parents) if (p / 'planting').exists())

COMBINE_DIR = PROJECT_ROOT / 'CombineAndCompare'
if str(COMBINE_DIR) not in sys.path:
    sys.path.insert(0, str(COMBINE_DIR))

from sensitivity_analysis import SCENARIOS, run_sensitivity_analysis

[(s.name, s.changed_input) for s in SCENARIOS]
```

## Experiment settings

The baseline is 500g initial cash, 900 minutes, 270 energy, and capacity for one tool in Blacksmith service. The `blacksmith_service_2` scenario changes only the service capacity; it keeps the daily submission capacity at one.

With eight scenarios and a 180-second cap per scenario, `Run All` may take up to about 24 minutes.

```python
TIME_LIMIT_SECONDS = 180
MIP_GAP_TARGET = 0.001

summary = run_sensitivity_analysis(
    PROJECT_ROOT,
    time_limit=TIME_LIMIT_SECONDS,
    mip_gap=MIP_GAP_TARGET,
)
summary.round(3)
```
